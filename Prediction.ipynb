```python
#Get the correlation
# Convert 'Attrition' column to numerical (Yes=1, No=0)
df_numeric = df.copy()
df_numeric['Attrition'] = df_numeric['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)

# Select only numeric columns for correlation
df_numeric = df_numeric.select_dtypes(include=[np.number])

# Calculate correlation
df_numeric.corr()
```

```python
#Visulaize the correlation
plt.figure(figsize=(14,14))
sns.heatmap(df_numeric.corr(),annot=True,fmt=".0%")
```

```python
#Transform the data
#Transform non-numerical to numerical
from sklearn.preprocessing import LabelEncoder

for column in df.columns:
    if df[column].dtype == 'object':

# Check if the column is of object type (i.e., categorical strings)
        df[column] = LabelEncoder().fit_transform(df[column])
```

```python
#Create a new column
df['Age_Year'] = df['Age']
```

```python
#Drop the age column as we have  a copy created already
df=df.drop('Age',axis=1)
```

```python
#Split the data
X = df.iloc[:, 1:df.shape[1]].values
Y = df.iloc[:, 0].values
```

```python
import sklearn
print(sklearn.__version__)
```

```python
#Split the data in 75% training and 25% testing
from sklearn.model_selection import train_test_split
X = pd.get_dummies(df.drop(columns="Attrition"), dtype=int).to_numpy()
Y = df["Attrition"].map({"No": 0, "Yes": 1}).to_numpy()

x_train, x_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.25, random_state=0, stratify=Y
)
```

```python
#Use the Random Forest Classifier
from sklearn.ensemble import RandomForestClassifier
forest = RandomForestClassifier(n_estimators=10, criterion='entropy', random_state=0)
forest.fit(x_train, y_train)
```

```python
#Get the accuracy on the traininig dataset
forest.score(x_train, y_train)
```

```python
#Show the confusion Matrix and the accuracy score for the model on the test data

from sklearn.metrics import confusion_matrix

CM = confusion_matrix(y_test, forest.predict(x_test))

TM = CM[0][0]
TP = CM[1][1]
FN = CM[1][0]
FP = CM[0][1]

print(CM)
print('Model Testing Accuracy = {}'.format((TP + TM) / (TP + TM + FN + FP)))
```

```python
# Return the feature importances (the higher, the more important the feature).
# Match feature names to the encoded training data (55 columns after one-hot encoding)
feature_names = pd.get_dummies(df.drop(columns="Attrition"), dtype=int).columns
importances = pd.DataFrame({
    'feature': feature_names,
    'importance': np.round(forest.feature_importances_, 3)
})

#Note: The target column is at position 0
importances = importances.sort_values('importance',ascending=False).set_index('feature')
importances
```

```python
#Visualize the importance
importances.plot.bar()
```
